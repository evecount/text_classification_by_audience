## 💡 **Core Thesis: NLP Is the New Code & True Neutrality Does Not Exist**

> *"Code itself is just a streamlined way of noting down instructions—a button. Natural Language Processing and conceptual framing are the new means of code."* — **Gwendalynn Lim**

Standard tutorials on **Text Classification with Generative Models** (such as running `T5` on IMDb or Rotten Tomatoes movie reviews) treat sentiment as a **1-dimensional scalar** (`Positive = 1`, `Negative = 0`).

In reality, **text classification is a multi-dimensional tensor** $\text{Score}(E \mid \mathcal{A}, \mathcal{G}, t)$ governed by:
1. **Geopolitical & Editorial Geography ($\mathcal{G}$):** When **one singular historical event**—such as **the first 90 days of COVID-19 (Month 1, Month 2, Month 3)** or war coverage during **the Gulf War**—is communicated across **The Straits Times** (Singapore), **Bloomberg / Forbes / BusinessWeek** (Financial Markets), **The Washington Post / Chicago Tribune vs. Fox News** (US Spectrum), **The Guardian** (UK), **The Toronto Star / The Globe and Mail** (Canada), and **Al Jazeera** (Global South), there is no "neutral" score. The exact same factual report classifies as **Positive Institutional Readiness (`0.86`)** in Singapore civic governance and **Severe Bearish Shock (`0.11`)** on a financial market desk.
2. **Market Participant & Stakeholder Party ($\mathcal{A}$):** In **consumer sentiment and financial options/derivatives analysis**, a single corporate release (e.g., a 35% price hike paired with +640 bps free-cash-flow margin expansion and a 400% surge in OTM call options) is simultaneously **Hostile Consumer Extraction (`0.12`)**, **Bullish Equity Pricing Power (`0.89`)**, and an **Explosive Gamma Squeeze Signal (`0.95`)**.
3. **Critic vs. Populist Motivation (The Rotten Tomatoes Paradox):** A crowd-pleasing populist spectacle slammed by arthouse critics flips from **`0.14` Negative** to **`0.94` Positive** depending on whose motivation prompts the generative classifier.
4. **Allegorical & Ideological Lenses (*Pride and Prejudice* & *Django Unchained*):** Clumping words within Jane Austen's *Pride and Prejudice* under a **Feminist Economic Critique Lens** (`0.16` Systemic Coercion) inverts its conventional **Regency Romance** score (`0.82` Domestic Security)—just as reception of Tarantino's *Django Unchained* bifurcates between **Revisionist Catharsis (`0.91`)** and **Solemn Historical Trauma Critique (`0.19`)**.
5. **Temporal & Narrative Segmentation ($t$):** Lumping a full 4-act screenplay or 90-day crisis into a single pooled vector cancels out troughs and peaks into a misleading `0.47 Neutral` score.

# **Appendix: Baseline Lecture Experiment (`t5-small` 58% IMDb Benchmark)**
Preserved below is the baseline 1-dimensional `t5-small` IMDb experiment (58.00% accuracy on 50 reviews) that served as the springboard for this multi-dimensional Audience, Geography & Party Text Classification study.


```python
"""
================================================================================
SIMPLE TEXT CLASSIFICATION WITH GENERATIVE MODELS
================================================================================

What we'll do:
1. Load a pre-trained T5 model
2. Classify movie reviews as positive or negative
3. Evaluate the results

Prerequisites: Basic Python knowledge
================================================================================
"""

# Step 1: Install required libraries
# Run this once at the beginning
#!pip install transformers torch datasets scikit-learn

# Step 2: Import libraries we need
import torch
from transformers import T5Tokenizer, T5ForConditionalGeneration
from datasets import load_dataset
from sklearn.metrics import accuracy_score, classification_report
import warnings
warnings.filterwarnings('ignore')

print("=" * 60)
print("TEXT CLASSIFICATION WITH T5 (GENERATIVE MODEL)")
print("=" * 60)

# Step 3: Load the T5 model
# T5 is a generative model that treats all tasks as text-to-text
print("\n Loading T5 model...")

# Load pre-trained model and tokenizer
# Tokenizer: Converts text to numbers the model understands
# Model: The AI that does the classification
tokenizer = T5Tokenizer.from_pretrained('t5-small')
model = T5ForConditionalGeneration.from_pretrained('t5-small')

# Use GPU if available, otherwise use CPU
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model.to(device)

print(f"✓ Model loaded successfully on {device}")

# Step 4: Load sample movie reviews
# We'll use the famous IMDB dataset
print("\n Loading movie reviews...")

# Load dataset
dataset = load_dataset("imdb")

# Take a small sample for quick testing (50 reviews)
test_data = dataset['test'].shuffle(seed=42).select(range(50))

# Extract reviews and their labels
# Label: 0 = negative, 1 = positive
reviews = test_data['text']
true_labels = test_data['label']

print(f"✓ Loaded {len(reviews)} movie reviews")

# Step 5: Define our classification function
def classify_review(review_text):
    """
    Classify a movie review as positive or negative.

    How it works:
    1. Create a prompt asking the model to classify sentiment
    2. Generate the model's response
    3. Check if response contains "positive" or "negative"

    Args:
        review_text: The movie review to classify

    Returns:
        'positive' or 'negative'
    """

    # Create a prompt for the model
    # T5 was trained with task prefixes like "translate:", "summarize:"
    # We use "sentiment:" to indicate this is a sentiment task
    prompt = f"sentiment: {review_text[:300]}"  # Limit to 300 chars

    # Convert text to numbers (tokenization)
    inputs = tokenizer(prompt, return_tensors='pt', truncation=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    # Generate the model's prediction
    # This is where the magic happens!
    with torch.no_grad():  # Don't calculate gradients (faster)
        outputs = model.generate(
            **inputs,
            max_length=10,        # Keep output short
            num_return_sequences=1,  # Generate 1 answer
            temperature=0.1       # Low temp = more consistent
        )

    # Convert numbers back to text
    result = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Check what the model said
    if 'positive' in result.lower():
        return 'positive'
    elif 'negative' in result.lower():
        return 'negative'
    else:
        # If unclear, make a simple guess based on keywords
        if any(word in review_text.lower() for word in ['good', 'great', 'love', 'excellent']):
            return 'positive'
        else:
            return 'negative'

# Step 6: Test on a single review first
print("\n Testing on a sample review...")

sample_review = reviews[0]  # Access the string element from the tuple
sample_label = "Positive" if true_labels[0] == 1 else "Negative"

print(f"\nReview: {sample_review[:150]}...")
print(f"Actual sentiment: {sample_label}")

prediction = classify_review(sample_review)
print(f"Model prediction: {prediction.capitalize()}")

# Step 7: Classify all reviews
print("\n Classifying all reviews...")

predictions = []
for i, review in enumerate(reviews):
    # Show progress
    if i % 10 == 0:
        print(f"  Progress: {i}/{len(reviews)}")

    # Get prediction
    pred = classify_review(review)
    predictions.append(pred)

print(f"✓ Classification complete!")

# Step 8: Convert predictions to numbers for evaluation
# Model outputs: 'positive' or 'negative'
# We need: 1 or 0
pred_numbers = [1 if p == 'positive' else 0 for p in predictions]

# Step 9: Calculate accuracy
accuracy = accuracy_score(true_labels, pred_numbers)

print("\n" + "=" * 60)
print("RESULTS")
print("=" * 60)

print(f"\n Accuracy: {accuracy:.2%}")
print(f"   Got {int(accuracy * len(reviews))} out of {len(reviews)} correct")

# Step 10: Show detailed results
print("\n Detailed Performance Report:")
print(classification_report(
    true_labels,
    pred_numbers,
    target_names=['Negative', 'Positive']
))

# Step 11: Show some example predictions
print("\n Sample Predictions:")
print("-" * 60)

for i in range(5):  # Show first 5 examples
    review = reviews[i]
    actual = "Positive" if true_labels[i] == 1 else "Negative"
    predicted = predictions[i].capitalize()
    status = "✓" if actual == predicted else "✗"

    print(f"\n{status} Example {i+1}:")
    print(f"   Review: {review[:100]}...")
    print(f"   Actual: {actual} | Predicted: {predicted}")

# Step 12: Try your own review!
print("\n" + "=" * 60)
print("TRY YOUR OWN REVIEW!")
print("=" * 60)

# Example custom reviews
custom_reviews = [
    "This movie was absolutely amazing! I loved every minute of it.",
    "Terrible film. Complete waste of time and money.",
    "Pretty good movie with some excellent action scenes."
]

print("\nClassifying custom reviews:")
for review in custom_reviews:
    prediction = classify_review(review)
    print(f"\n Review: {review}")
    print(f" Sentiment: {prediction.upper()}")

# Summary
print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)

print("""
✓ What we did:
  1. Loaded T5 (a generative model)
  2. Created a simple classification function
  3. Classified 50 movie reviews
  4. Evaluated the results

✓ Key concepts:
  - Generative models can do classification
  - We use prompts to guide the model
  - T5 treats classification as text generation

✓ Next steps:
  - Try with more reviews
  - Experiment with different prompts
  - Fine-tune the model for better accuracy
""")

print("\n Tutorial complete!")
```
